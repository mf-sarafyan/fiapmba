```python

url = "https://raw.githubusercontent.com/diogenesjusto/FIAP/master/dados/credit.csv"

def ler_credit_csv(url):
    """Tenta ler o CSV por UTF-8; se falhar, tenta Latin-1; se ainda falhar, CP1252 com tolerância."""
    try:
        df = pd.read_csv(url)  # tenta UTF-8
        print("Leitura OK com UTF-8")
        return df, "utf-8"
    except UnicodeDecodeError:
        print("UTF-8 falhou; tentando Latin-1...")
        try:
            df = pd.read_csv(url, encoding="latin1", sep=",")
            print("Leitura OK com Latin-1")
            return df, "latin1"
        except UnicodeDecodeError:
            print("Latin-1 falhou; tentando CP1252 com tolerância...")
            df = pd.read_csv(url, encoding="cp1252", sep=",",
                             encoding_errors="replace", on_bad_lines="skip")
            print("Leitura OK com CP1252 (replace/skip)")
            return df, "cp1252"

df, enc_usado = ler_credit_csv(url)
print("Encoding usado:", enc_usado)
print("Dimensão:", df.shape)

df.head()
```
